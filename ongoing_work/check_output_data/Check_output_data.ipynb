```python
import numpy as np
import scipy.io as sio
import pandas as pd
```

```python
full = sio.loadmat(r"C:\Users\maula\Desktop\PhD FIles\Codes\DDC_python\combined_Simulation_2_dark_state_sparse_Clusters_10per_3_per_ROI.mat", squeeze_me=True, struct_as_record=False)
cut = sio.loadmat(r"C:\Users\maula\Desktop\PhD FIles\Codes\DDC_python\combined_cut_firstimage_firstrun.mat", squeeze_me=True, struct_as_record=False)
cut2 = sio.loadmat(r"C:\Users\maula\Desktop\PhD FIles\Codes\DDC_python\combined_cut_firstimage_secondrun.mat", squeeze_me=True, struct_as_record=False)
cut3 = sio.loadmat(r"C:\Users\maula\Desktop\PhD FIles\Codes\DDC_python\combined_cut_firstimage_thirdrun.mat", squeeze_me=True, struct_as_record=False)
```

```python
# Compare two arrays if they are equal
def compare_floating_point_arrays(arr1, arr2, atol=1e-6):
    arr1 = np.array(arr1, dtype=float)
    arr2 = np.array(arr2, dtype=float)

    matched_indices_2 = set()
    data = []

    for i, p1 in enumerate(arr1):
        found = False
        for j, p2 in enumerate(arr2):
            if j in matched_indices_2:
                continue
            if np.allclose(p1, p2, atol=atol):
                data.append([*p1, *p2, True])
                matched_indices_2.add(j)
                found = True
                break
        if not found:
            data.append([*p1, np.nan, np.nan, np.nan, False])

    # Add unmatched points in arr2
    for j, p2 in enumerate(arr2):
        if j not in matched_indices_2:
            data.append([np.nan, np.nan, np.nan, *p2, False])

    df = pd.DataFrame(data, columns=["x1", "y1", "z1", "x2", "y2", "z2", "similar"])
    return df

print(full['Final_Loc_Blinking_Corrected'][0].shape)
print(cut['Final_Loc_Blinking_Corrected'].shape)
print(cut2['Final_Loc_Blinking_Corrected'].shape)

df1 = compare_floating_point_arrays(full['Final_Loc_Blinking_Corrected'][0], cut['Final_Loc_Blinking_Corrected'])
df2 = compare_floating_point_arrays(full['Final_Loc_Blinking_Corrected'][0], cut2['Final_Loc_Blinking_Corrected'])
df3 = compare_floating_point_arrays(cut['Final_Loc_Blinking_Corrected'], cut2['Final_Loc_Blinking_Corrected'])
df4 = compare_floating_point_arrays(cut['Final_Loc_Blinking_Corrected'], cut3['Final_Loc_Blinking_Corrected'])
df5 = compare_floating_point_arrays(cut2['Final_Loc_Blinking_Corrected'], cut3['Final_Loc_Blinking_Corrected'])
```

```python
def similar_ratio(df):
    similar_count = df['similar'].sum()
    total_count = len(df)
    return similar_count / total_count if total_count > 0 else 0

print("Similarity ratio between full and cut:", similar_ratio(df1))
print("Similarity ratio between full and cut2:", similar_ratio(df2))    
print("Similarity ratio between cut and cut2:", similar_ratio(df3))
print("Similarity ratio between cut and cut3:", similar_ratio(df4))
print("Similarity ratio between cut2 and cut3:", similar_ratio(df5))
```
